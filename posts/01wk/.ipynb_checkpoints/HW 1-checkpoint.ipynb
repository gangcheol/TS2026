## sol

`1` ref

$$P(E) =  \sum_{i=1}^n P(E|F_{i})\times P(F_{i})$$

$$\text{where } F_{i}\,\cap\, F_{j} = \varnothing,\,\, \text{for all } i \neq j \text{ and} \cup_{i=i}^{n} F_{i}=S$$ 

`2` 문제 적용

1. $P(A)$: Accident-probe(사고다발자) = `0.3`

2. $P(A^{c})$: Non-accident-prone(비사고다발자) = `0.7`

3. $P(A_{1}|A)$: 사고다발자가 1년 내 사고를 낼 확률 = `0.4`

4. $P(A_{1}|A^{c})$: 사고다발자가 1년 내 사고를 낼 확률 = `0.2`

$\divideontimes \,\, A \cap A^{c} = \varnothing, \,\, A \cup A^{c} = S$

`3` 따라서

$$\begin{align*}P(A_{1}) &= P(A_1|A)P(A) + P(A_1|A^{c})P(A^{c}) \\ \\ 
                         &= (0.4 \times 0.3) + (0.2\times0.7) = 0.26\end{align*}$$

$$\begin {align*}P(A|A_{1}) &=  \frac{P(A \cap A_{1})}{P(A_{1})} \\ \\ 
                            &= \frac{P(A_{1}|A)\times P(A)}{P(A_{1})} = \frac {0.3 \times 0.4}{0.26} = \frac{6}{13} \end{align*}$$

# example related to two fair dice on page 14

<center><img src = 'ex2.jpg' width = 700></center>

## sol

`1` 각 확률 계산

* $P(E_{7})$은 두 개의 주사위 합이 7인 경우 이므로 $\to$ (1,6), (2,5), (3,4), (4,3), (5,2), (6,1)

$$P(E_{7}) = \frac{6}{36} = \frac{1}{6}$$

* $P(F)$은 첫번째 주사위가 4가 나올 확률이므로

$$P(F) = \frac{6}{36} = \frac{1}{6}$$

* $P(T)$은 두번째 주사위가 3가 나올 확률이므로

$$P(T) = \frac{6}{36} = \frac{1}{6}$$

`2` Pairwise indep check

1. $E_{7}, F$: 두 개의 주사위 합이 7이면서, 첫 번째 주사위가 4가 나올 확률
    - 이 경우는 (4,3) 밖에 없음.

$$\therefore \,\,P(E_{7} \cap F) = \frac{1}{36} = P(E_{7}) \times P(F)$$

2. 이 이후에 pairwise도 `1/36`로 동일하며 `pairwise indep`를 보임.

`3` $P(E_{7}\cap F \cap T) = P(E_7)P(F)P(T)$?

- $P(E_{7}\cap F \cap T)$: 두 주사위의 합이 7이면서, 첫번째 주사위의 눈이 4 두번째 주사위의 눈이 3일 확률.

$$\therefore \,\, P(E_{7}\cap F \cap T) = \frac{1}{36}$$

* but..... $P(E_7)P(F)P(T) = (\frac{1}{6})^{3} = \frac{1}{216}$

$$\therefore \,\, P(E_{7}\cap F \cap T) \neq P(E_7)P(F)P(T)$$

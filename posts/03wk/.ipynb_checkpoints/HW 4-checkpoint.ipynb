## ex1

Suppose that $X_{1} \dots X_n$ is an iid sample from $U(0, \theta)$

(a) Find the limiting distribution of $X_{(n)}$

(b) Find the limiting distribution of $Y_{n} = n(\theta - X_{(n)})$

## sol

### (a)

$$P(|X_{(n)}-\theta| > \epsilon) = P(X_{(n)} \leq \theta - \epsilon) = \left ( \frac {\theta - \epsilon}{\epsilon}\right )^{n} \to 0$$

$$\divideontimes\,\, X_{(n)} \xrightarrow{d} \theta$$

### (b)

$$P(Y_n > y) =  P(n(\theta - X_{(n)}) >y) =  P(X_{(n)} > \theta - \frac yn) = \left (1 - \frac{y}{n\theta}\right )^{n} \to e^{-y/\theta}$$

$$P(Y_{n}  \geq y) \to 1 - e^{-y/\theta} \quad (y \geq 0)$$

$$Y_{n}  \sim \text{exp}(\lambda = 1/\theta), \quad \theta \to \infty$$

# ex2

Suppose that $X_1 \dots X_n$ is an idd sample from  $f_{X}(x) = e^{-(x-\mu)}I(x > \mu)$

(a) Show that the probability density function of the minimum order statistic $X_{(1)}$ is $f_{x_{(1)}}(x) = ne^{-n(x-\mu)}I(x > \mu)$

(b) Show that $X_{(1)}$ converges in probability to $\mu$

## sol

### (a)

`step1`

$$F(x)  = \int_{\mu}^{x} f(t) \,dt =   1-e^{-(x-\mu)}$$

$$P(X>x) = 1- F(x) =e^{-(x-\mu)}$$

`step2`

$$P(X_{(1)}>x) =  P(X_{1}>x,\dots X_{n}>x)$$

$$P(X_{(1)}>x) =  \{P(X>x)\}^{n} = e^{-n(x-\mu)}$$

$$F_{X_{(1)}}(x) = 1-P(X_{(1)}>x) = 1-e^{-n(x-\mu)}$$

`step3`

$$\divideontimes \,\,f_{x_{(1)}}(x) = \frac{d}{dx} F_{x_{(1)}}(x) = ne^{-n(x-\mu)}$$

### (b)

$$P(|X_{(1)} - \mu| > \epsilon) = P(X_{(1)}> \mu + \epsilon) = e^{-nx}$$

$$n\to \infty,\,\, e^{-nx} \to 0$$

$$\divideontimes\,\, P(|X_{(1)} - \mu| > \epsilon) = P(X_{(1)}> \mu + \epsilon) = e^{-nx} \to 0$$

$$\divideontimes\,\, X_{(1)} \xrightarrow{P} \mu$$

# ex3

Suppose that $X_1 \dots X_n$ is an idd sample from  $\text{Pareto} (\alpha, \beta), \,\, \text{where  } \alpha>0, \,\,\beta >0$

Recall that the Pareto probability  density function is given by

$$f_{X}(x)= \frac {\beta\alpha^{\beta}}{x^{\beta+1}} I(x>\alpha)$$

Let $X_{(1)}$ and $X_{(n)}$ denote the minimum and maximum order statistics, respectively, 

(a) Show that $X_{(1)}  \sim \text{Pareto}(\alpha, n\beta)$.

(b) Show that cumulative distribution function of $X_{(n)}$ is

$$F_{X_{(n)}} (x) = \left [1- \left (\frac{\alpha}{x} \right )^{\beta}\right ]^{n}, \quad x>\alpha $$

(c) Define the sequence of real numbers $b_n = \alpha n^{1/\beta},$ for $n= 1\dots$  Determine

$$\lim_{n \to \infty} F_{X(n)}(b_{n}x)$$

This limit turns out to be the cdf of the standard Frechet distribution, a useful model in extreme value theory

## sol

### (a)

`step1`

$$F(x) = 1-\left( \frac{\alpha}{x}\right)^\beta$$

$$P(X_{(1)}>x) = \{ P(X>x)\}^n = (1-F(x))^{n} = \left( \frac{\alpha}{x}\right)^{n\beta } $$

`step2`

$$F_{X_{(1)}}(x) = P(X_{(1)} \leq x) = 1- P(X_{(1)} > x) \begin{cases} 0, \quad x\leq \alpha \\ \\ 
                                                                        1- \left( \frac{\alpha}{x}\right)^{n\beta }, \quad x>\alpha. \end{cases}$$

`step3`

$x>\alpha$에서

$$f_{X_{(1)}}(x) = \frac{d}{dx}\,\,\left [  1- \left( \frac{\alpha}{x}\right)^{n\beta }\right ] = n\beta \alpha^{n\beta}x^{-(n\beta + 1)} $$

$$\divideontimes \,\, f_{x_{(1)}} (x) = \frac { n\beta \alpha^{n\beta}}{x^{n\beta + 1}}, \quad X_{(1)} \sim \text{Pareto}(\alpha, n\beta)$$

### (b)

$$F(x) = 1-\left( \frac{\alpha}{x}\right)^\beta$$

이므로

$$F_{X_{n}}(x) = P(X_1 \leq x\dots X_n \leq x) = \{ P(X\leq x)\}^{n} = [F(x)]^{n}$$

따라서

$$F_{X_{(n)}} (x) = \left [ 1-\left( \frac{\alpha}{x}\right)^\beta \right ]^{n}, \quad x > \alpha$$

----

### (c)

$$F_{X_{(n)}} (x) = \left [ 1-\left( \frac{\alpha}{x}\right)^\beta \right ]^{n}, \quad x > \alpha$$

이므로

$$F_{X_{(n)}} (b_{n}x) = \left [ 1-\left( \frac{\alpha}{b_{n}x}\right)^\beta \right ]^{n}, \quad x > \alpha$$

그런데

$$\left( \frac{\alpha}{b_{n}x}\right)^\beta =  \frac{1}{n^{x\beta}}$$

따라서

$$F_{X_{(n)}} (b_{n}x) = \left [ 1-\frac{1}{n^{x\beta}} \right ]^{n}, \quad x > \alpha$$

$$\lim_{n\to\infty} F_{X_{(n)}}(b_n x)=\begin{cases}0, & x \le 0,\\[4pt] \exp\!\left(-x^{-\beta}\right), & x>0. \end{cases}$$

Frechet 분포를 따른다.

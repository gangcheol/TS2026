- Gamma function quick facts ($\Gamma$)

$$\Gamma(\alpha) = \int_{0}^{\infty} x^{\alpha-1}e^{-x}\,dx $$

$$\Gamma(\alpha + 1) = \alpha\Gamma(\alpha),\quad \Gamma(n) = (n-1)!, \quad \Gamma(1/2) = \pi $$

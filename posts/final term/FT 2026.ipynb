# 2번

Suppose $X_1, X_2 \dots X_n$ are iid with continuous cdf $F$ and pdf $f$. Find the limiting distribution of the sample median $\hat{\phi}_{0.5}$

## Sol

`1` 표본 분위수의 점근 분포

$$\sqrt{n}(\hat{\phi}_p - \phi_p) \xrightarrow{d} \mathcal{N}\left(0, \frac{p(1-p)}{f^2(\phi_p)}\right)$$

`2` 중앙값의 조건대입

$$p = 0.5$$

$$\frac{p(1-p)}{f^2(\phi_p)} = \frac{0.5(1-0.5)}{f^2(\phi_{0.5})} = \frac{0.25}{f^2(\phi_{0.5})} = \frac{1}{4f^2(\phi_{0.5})}$$

`3` 최종 점근 분포

$$\sqrt{n}(\hat{\phi}_{0.5} - \phi_{0.5}) \xrightarrow{d} \mathcal{N}\left(0, \frac{1}{4f^2(\phi_{0.5})}\right)$$

$$\divideontimes\,\, \hat{\phi}_{0.5} \approx \mathcal{N}\left(\phi_{0.5}, \frac{1}{4nf^2(\phi_{0.5})}\right)$$

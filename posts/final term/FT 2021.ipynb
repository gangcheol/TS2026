# 1번

$X_1, X_2, ..., X_n$ 은 독립동일분포(i.i.d)로 Bernoulli($\theta$)를 따르며, 확률질량함수는 $f(x|\theta)=\theta^{x}(1-\theta)^{1-x}$ (단, $x=0,1$)입니다.

## (a)

$\theta$의 피셔 정보수 $I(\theta)$ 

### sol

`1` 1차 미분

$$\frac{\partial}{\partial \theta} \log f(x|\theta) = \frac{x}{\theta} - \frac{1-x}{1-\theta}$$

`2` 2차 미분

$$\frac{\partial^2}{\partial \theta^2} \log f(x|\theta) = -\frac{x}{\theta^2} - \frac{1-x}{(1-\theta)^2}$$

`3` 피셔의 정보수

$$\begin{align*}I(\theta) &= -E\left[ \frac{\partial^2}{\partial \theta^2} \log f(X|\theta) \right]\\ \\
&= -\left( -\frac{\theta}{\theta^2} - \frac{1-\theta}{(1-\theta)^2} \right)  \\ \\ &= \frac{1}{\theta} + \frac{1}{1-\theta} = \frac{1}{\theta(1-\theta)} \end{align*}$$

## (b)

$\theta(1-\theta)$의 최대 가능도 추정량(MLE)과 평균제곱오차(MSE)

### sol

`1` $\theta$의 MLE는 표본평균 $\hat{\theta} = \bar{X}$ 

`2` MLE의 불변성(Invariance property)에 의해 $\theta(1-\theta)$의 MLE는 $\bar{X}(1-\bar{X})$

`3` MSE 구하기

- bias 구하기: $\text{bias} = E[\theta]-\theta$

$$E[\bar{X}^2] = Var(\bar{X}) + (E[\bar{X}])^2 = \frac{\theta(1-\theta)}{n} + \theta^2$$

$$\begin{align*}E[\bar{X}(1-\bar{X})] &= \theta - \left( \frac{\theta(1-\theta)}{n} + \theta^2 \right) \\ \\ 
&= \theta(1-\theta) - \frac{\theta(1-\theta)}{n} \\ \\ &= \frac{n-1}{n}\theta(1-\theta)\end{align*}$$

- 분산 구하기: delta metthod 활용

$$g(\theta) = \theta(1-\theta), \quad g'(\theta) = 1-2\theta$$

$$Var(g(\hat{\theta})) \approx (g'(\theta))^2 Var(\hat{\theta}) = (1-2\theta)^2 \frac{\theta(1-\theta)}{n}$$

`4` 최종 MSE: Var + $\text{bias}^2$

$$MSE \approx (1-2\theta)^2 \frac{\theta(1-\theta)}{n} + \left( -\frac{\theta(1-\theta)}{n} \right)^2$$

## (c)

$\theta(1-\theta)$의 최대 가능도 추정량의 극한분포

### sol

`1` 중심극한정리

$$\sqrt{n}(\bar{X} - \theta) \xrightarrow{d} N(0, \theta(1-\theta))$$

`2` 델타방법 적용

$$\sqrt{n}(g(\bar{X}) - g(\theta)) \xrightarrow{d} N(0, [g'(\theta)]^2 \theta(1-\theta))$$

`3` 대입

$$\sqrt{n}(\bar{X}(1-\bar{X}) - \theta(1-\theta)) \xrightarrow{d} N\left(0, (1-2\theta)^2 \theta(1-\theta)\right)$$

## (d)

$\theta(1-\theta)$에 대한 근사적인 95% 신뢰구간

### sol

`1` (c)의 극한분포 결과를 바탕으로 표준오차는 다음과 같이 계산됨 

$$SE = \sqrt{\frac{(1-2\hat{\theta})^2 \hat{\theta}(1-\hat{\theta})}{n}}$$

`2` 따라서, 95% 신뢰구간은

$$\bar{X}(1-\bar{X}) \pm 1.96 \times SE$$

## (e)

### sol

$\theta(1-\theta)$의 최소분산불편추정량(UMVUE)

`1` 레만-셰페(Lehmann-Scheffe) 정리 적용

- 지수족 C.S.S 및 비편향 추정량 확인

`2` $T = \sum X_i$ 는 지수족에 속하므로 $\theta$에 대한 완비충분통계량임

`3` (b)에서 $E[\bar{X}(1-\bar{X})] = \frac{n-1}{n}\theta(1-\theta)$ 임을 확인 

`4` 따라서

$$\text{UMVUE} = \frac{n}{n-1}\bar{X}(1-\bar{X})$$

# 2번

$X\sim N(\theta,1), Y\sim N(\theta,4)$ 이며 독립입니다

## (a)

$T_{a}=aX+(1-a)Y$ 가 $\theta$의 비편향추정량임 증명 

### sol

$$\begin{align*} E[T_a] &= E[aX + (1-a)Y] \\ \\ &= aE[X] + (1-a)E[Y] \\ \\ &= a\theta + (1-a)\theta = \theta \end{align*}$$

## (b)

분산을 최소로 하는 상수 $a$ 

### sol

$$\begin{align*}Var(T_a)  &= a^2Var(X) + (1-a)^2Var(Y) \\ \\ &= a^2(1) + (1-a)^2(4) \\ \\ &= 5a^2 - 8a + 4\end{align*}$$

$$\frac{d}{da}(5a^2 - 8a + 4) = 10a - 8 = 0 \Rightarrow a = \frac{4}{5} = 0.8$$

## (c)

$H_{0}:\theta=0 \text{ vs } H_{1}:\theta=-2$ 에 대한 최강력 검정법

### sol

`1` 네이만-피어슨 보조정리(Neyman-Pearson Lemma)를 이용합니다.

`2` 우도비

$$LR = \frac{L(-2)}{L(0)} = \frac{\exp(-\frac{1}{2}(x+2)^2 - \frac{1}{8}(y+2)^2)}{\exp(-\frac{1}{2}x^2 - \frac{1}{8}y^2)}$$

`3` 로그 우도비

$$\begin{align*}\ln LR &= -\frac{1}{2}(x^2+4x+4 - x^2) - \frac{1}{8}(y^2+4y+4 - y^2) \\ \\ &= -2x - 2 - \frac{1}{2}y - \frac{1}{2} \\ \\ &= -2x - \frac{1}{2}y - \frac{5}{2}\end{align*}$$

`4` $H_0$ 기각 조건은 $\ln LR > k'$ 이므로

$$-2x - \frac{1}{2}y > C \Rightarrow 4x + y < C'$$

`5` 통계량 정의  $W = 4X + Y$

$$W \sim N(0, 4^2(1) + 1^2(4)) = N(0, 20)$$

`6` 유의수준 $\alpha$에서 아래와 같으므로

$$P(4X + Y < c | H_0) = \alpha$$

`7` 최강력 기각역을 아래와 같이 산출

$$\frac{c}{\sqrt{20}} = -z_\alpha$$

$$4X + Y < -\sqrt{20}z_\alpha$$

# 3번

$X_1, ..., X_{25} \sim N(\theta, 1)$ 표본에 대해 $H_0: \theta=9 \text{ vs } H_1: \theta=10$ 이고 기각역이 $\{\bar{X} > 9.5\}$이다.

유의수준과 검정력을 구하시오.

## sol

`1` 표준편차는 0.2

$\bar{X} \sim N(\theta, \frac{1}{25})$

`2` 유의수준 $\alpha$ 구하기

$$\begin{align*} \alpha &= P(\bar{X} > 9.5 | \theta=9) \\ \\ &= P\left(Z > \frac{9.5 - 9}{0.2}\right) \\ \\ &= P(Z > 2.5) = 1 - \Phi(2.5)\end{align*}$$

`3` 검정력

$$\begin{align*}\text{검정력} &= P(\bar{X} > 9.5 | \theta=10) \\ \\ &= P\left(Z > \frac{9.5 - 10}{0.2}\right) \\ \\ &= P(Z > -2.5) = 1 - \Phi(-2.5) = \Phi(2.5)\end{align*}$$

# 4번

$X_1, ..., X_n \sim Exp(\theta)$ ($f(x|\theta)=\frac{1}{\theta}e^{-x/\theta}$)에 대하여 $H_0: \theta=\theta_0 \text{ vs } H_1: \theta \ne \theta_0$ 를 검정

## (a)

전체 모수공간 $\Omega$ 와 귀무가설 하의 모수공간 $\Omega_0$ 

### sol

$\Omega = \{ \theta \mid \theta > 0 \}$, $\Omega_0 = \{ \theta_0 \}$

## (b)

우도함수(가능도함수)

### sol

$$L(\theta) = \prod_{i=1}^n \frac{1}{\theta} e^{-x_i/\theta} = \theta^{-n} \exp\left(-\frac{\sum x_i}{\theta}\right)$$

## (c)

$\Omega$ 에서의 MLE와 $\Omega_0$ 에서의 MLE 

### sol

`1` $\Omega$에서의 MLE

$$\log L(\theta) = -n\log\theta - \frac{\sum X_i}{\theta}, \quad \hat{\theta} = \frac{\sum X_i}{n} = \bar{X}$$

`2` $\Omega_0$에서의 MLE

 $\Omega_0$는 단일점이므로 $\hat{\theta}_0 = \theta_0$

## (d)

가능도비 $\Lambda$ 

### sol

$$\Lambda = \frac{\sup_{\Omega_0} L(\theta)}{\sup_{\Omega} L(\theta)} = \frac{L(\theta_0)}{L(\bar{X})} = \frac{\theta_0^{-n} \exp\left(-\frac{\sum X_i}{\theta_0}\right)}{\bar{X}^{-n} \exp\left(-\frac{\sum X_i}{\bar{X}}\right)} = \left( \frac{\bar{X}}{\theta_0} \right)^n \exp\left( n - \frac{n\bar{X}}{\theta_0} \right)$$

## (e)

유의수준 α인 가능도비검정법의 기각역 (기각역을 $\chi^2$분포의 분위수를 사용하여 표현)

### sol

`1` 검정통계량

$$-2\ln\Lambda = 2n\left[ \frac{\bar{X}}{\theta_0} - 1 - \ln\left(\frac{\bar{X}}{\theta_0}\right) \right]$$

`2` 기각역

$$-2\ln\Lambda > \chi^2_{1, \alpha}$$

- $\chi^2_{1, \alpha}$ 자유도가 1인 카이제곱 분포의 우측 상위 $\alpha$ 백분위수

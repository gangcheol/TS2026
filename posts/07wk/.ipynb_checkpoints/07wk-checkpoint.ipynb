# Metho of Moments

`1` $ E(X^{j})= m_j^{\prime} = \frac{1}{n} \sum_{i=1}^{n} X_i^{j}$: jth sample moment

`2`  즉, `jth`차 샘플 적률을 활용해서 $\theta$를 추정하는 것

## ex1

$X \sim \text{Uniform}(0, \theta)$

###  sol

$E(X) = \theta/2$

-> $\hat\theta_{mom} =2 \bar X$

## ex2

$X \sim \text{Uniform}(-\theta, \theta)$

$E(X) = 0, \,\,E(X^2) = \theta^{2} / 3$

$\divideontimes \quad \hat\theta_{mom} = \sqrt{3\bar {X^{2}} }$

## ex3

$X \sim N( \mu, \sigma^2)$

$\hat \mu_{mom} = \bar X$

$\sigma_{mom} = d$

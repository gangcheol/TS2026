# EX 1

Suppose that $X_1, X_2, . . . , X_n$ is an iid sample from 

$$f_{X}(x|\theta) = 2\theta x e^{-\theta x^2}\,\, I(x>0)$$ 

(a) Find the uniformly minimum variance unbiased estimator (UMVUE) of $\theta$

(b) Find the uniformaly most powerful(UMP) level $\alpha$ test of

$$H_0: \theta \leq \theta_0 \quad \text{vs} \quad H_1: \theta > \theta_0$$

where $\theta_0$ is known$

(c) Derive an expression for the power function of the test in part (b).

## sol

### (a)

`1`  지수족 인수분해를 통한 완비충분통계량 도출 $\to \,\, T = \sum X_{i}^{2}$

`2` 변수변환을 통해 $X_{i}^{2} = \text{Exp}(\theta) = \Gamma (1, 1/\theta)$ 이므로, $T \sim \Gamma(n, 1/\theta)$

`3` 불편성 조건 확인: $E[1/T] = \frac{\theta}{n-1} \to E\left [\frac{n-1}{T}\right] = \theta$

$\divideontimes \,\,\widehat{\theta}_{\text{UMVUE}} = \frac{n-1}{\sum_{i=1}^n X_i^2}$

### (b)

`1` 가능도비 $\frac{L(\theta_2)}{L(\theta_1)}$가 $T = \sum_{i=1}^n X_i^2$에 대해 감소하므로, 단조가능도비(MLR) 성질에 의해 기각역은 $Sub(T) < c$ 형태.

`2` $\theta = \theta_0$일 때 $2\theta_0 T \sim \chi^2_{2n}$임을 이용하여 크기 $\alpha$ 맞춤.

$\divideontimes \,\, \sum_{i=1}^n X_i^2 < \frac{\chi^2_{2n, 1-\alpha}}{2\theta_0}$ 이면 $H_0$ 기각 ($\chi^2_{2n, 1-\alpha}$는 카이제곱분포의 하위 $\alpha$ 백분위수)

### (c)

`1` 임의의 $\theta$에서 기각 확률 계산: $\beta(\theta) = P_\theta\left(\sum_{i=1}^n X_i^2 < \frac{\chi^2_{2n, 1-\alpha}}{2\theta_0}\right)$

`2` 양변에 $2\theta$를 곱해 카이제곱 통계량 $2\theta T \sim \chi^2_{2n}$ 도입.

`3` $\divideontimes \,\,\beta(\theta) = F_{\chi^2_{2n}}\left(\frac{\theta}{\theta_0} \chi^2_{2n, 1-\alpha}\right)$ ($F_{\chi^2_{2n}}$은 자유도 $2n$인 카이제곱 cdf)

# EX 2

Suppose that $X_1, X_2, . . . , X_n$ is an iid sample from 

$$f_{X}(x|\theta) = \frac{a_0}{\theta^{a_0}} x^{a_0 -1} \,\,I(0<x<\theta)$$

where $a_0 \geq 1$  is a known constant and $\theta > 0$ is an unknown parameter.

(a) Show that $Q = X_{(n)}/\theta$ is a pivotal quantity, where $X_{(n)}$ is the maximum order statistic. Use $Q$to construct a $1 − \alpha$ confidence set for $\theta$

(b)  Derive a $1 − \alpha$ confidence set for $\theta$ by using the cumulative distribution function of $T = T(X) = X_{(n)}$ as a pivot.

(c) Assume that has an inverse gamma prior distribution with parameters $c > 0$ and $d > 0$ (both known). Write a $1 − \alpha$ credible set for $\theta$.

## sol

### (a)

`1` $Q$의 cdf: $F_Q(q) = P\left(\frac{X_{(n)}}{\theta} \le q\right) = q^{n a_0}$ ($0<q<1$). 모수 $\theta$가 없으므로 축량(Pivot)임.   

`2` 가장 짧은 구간을 위해 $P(\alpha^{1/n a_0} \le Q \le 1) = 1-\alpha$로 설정 후 $\theta$에 대해 정리.

$\divideontimes\,\,\left[ X_{(n)}, \;\frac{X_{(n)}}{\alpha^{\frac{1}{n a_0}}} \right]$

### (b)

`1` 확률적분변환(PIT)에 의해 $U = F_T(X_{(n)}|\theta) = \left(\frac{X_{(n)}}{\theta}\right)^{n a_0} \sim \text{Uniform}(0,1)$은 축량임.

`2` $P(\alpha \le U \le 1) = 1-\alpha$ 대입 후 식을 정리하면 (a)와 동일한 식에 도달.

`3` $\divideontimes \,\,\left[ X_{(n)}, \;\frac{X_{(n)}}{\alpha^{\frac{1}{n a_0}}} \right]$

### (c)

`1` 가능도함수 $L(\theta|\mathbf{x}) \propto \theta^{-n a_0} I(\theta > X_{(n)})$와 사전분포 $\pi(\theta) \propto \theta^{-(c+1)} e^{-d/\theta}$ 곱함. 

`2` 사후분포: $\pi(\theta|\mathbf{x}) \propto \theta^{-(c + n a_0 + 1)} e^{-d/\theta} I(\theta > X_{(n)})$ (즉, 좌측 절단된 역감마 분포).

`3` 균등 꼬리(Equal-tail) 신용구간 $[k_1, k_2]$는 사후 pdf의 적분 조건식으로 정의됨.

`4` 아래 조건을 만족하는 구간 $[k_1, k_2]$

$$\frac{\Gamma\left(c+n a_0, \frac{d}{k_2}\right) - \Gamma\left(c+n a_0, \frac{d}{k_1}\right)}{\gamma\left(c+n a_0, \frac{d}{X_{(n)}}\right)} = 1 - \alpha \quad (\text{단, } X_{(n)} \le k_1 < k_2)$$

(여기서 $\gamma$는 Lower, $\Gamma$는 Upper 불완전 감마 함수)

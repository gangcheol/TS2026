# EX 1

Suppose that $X_1, X_2, . . . , X_n$ is an iid sample from 

$$f_{X}(x|\theta) = 2\theta x e^{-\theta x^2}\,\, I(x>0)$$ 

(a) Find the uniformly minimum variance unbiased estimator (UMVUE) of $\theta$

(b) Find the uniformaly most powerful(UMP) level $\alpha$ test of

$$H_0: \theta \leq \theta_0 \quad \text{vs} \quad H_1: \theta > \theta_0$$

where $\theta_0$ is known

(c) Derive an expression for the power function of the test in part (b).

- 즉, 검정법의 파워를 구하라

## sol

### (a)

`1`  지수족 인수분해를 통한 완비충분통계량 도출 $\to \,\, T = \sum X_{i}^{2}$

`2` 변수변환을 통해 $X_{i}^{2} = \text{Exp}(\theta) = \Gamma (1, 1/\theta)$ 이므로, $T \sim \Gamma(n, 1/\theta)$

`3` 불편성 조건 확인: $E[1/T] = \frac{\theta}{n-1} \to E\left [\frac{n-1}{T}\right] = \theta$

$\divideontimes \,\,\widehat{\theta}_{\text{UMVUE}} = \frac{n-1}{\sum_{i=1}^n X_i^2}$

### (b)

`0` most powerful test 찾기

$$H_0: \theta = \theta_0 \quad \text{vs} \quad H_1: \theta = \theta_1\,\,(\theta_1 > \theta_0)$$

`1` 기각역 정의

$$R = \left \{ (x_1 \dots x_n) : \frac{f_{\boldsymbol{X}(\boldsymbol{x}|\theta_1)}}{f_{\boldsymbol{X}(\boldsymbol{x}|\theta_0)}}\geq K \right \}$$

`2` 기각역 공식 대입

$$R = \left \{ (x_1 \dots x_n) : \frac{f_{\boldsymbol{X}(\boldsymbol{x}|\theta_1)}}{f_{\boldsymbol{X}(\boldsymbol{x}|\theta_0)}}
                = \frac{\prod_{i=1}^n 2\theta_1 x_i e^{-\theta_1 x_i^2}}{\prod_{i=1}^n 2\theta_0 x_i e^{-\theta_0 x_i^2}} =
                 \left (\frac{\theta_1}{\theta_0}\right)^n e^{-(\theta_1-\theta_0)\sum x_i^2}\geq K \right \}$$

`3` 단순화

$$R = \left \{ \sum x_i^2 \leq K \right \},\quad (\theta_1 > \theta_0)$$

`4` $\sum x^2 \sim \text{Gam}\left(n, \frac{1}{\theta} \right) \to 2\theta \sum x^2 \sim \text{Gam}\left (n,2\right) = \chi^2(2n)$

$$P\left [2\theta_0\sum x_i^2 \leq K\,|\, \theta = \theta_0 \right] = P\left [2\theta_0\sum x_i^2 \leq \chi^2_{1-\alpha}(2n)\,|\, \theta = \theta_0 \right] $$

- size(level-$\alpha$)에서 검정력 power는 다음과 같음

$$\beta(\theta) = P\left [2\theta_0\sum x_i^2 \leq \chi^2_{1-\alpha}(2n) \right]$$

`5` 균일최강력검정 구하기

$$\begin{align*}\max_{\theta \leq \theta_0}&=\max_{\theta \leq \theta_0}P\left[2\theta_0 \sum_{i=1}^{n} x_{i}^2 \leq \chi^2_{1-\alpha}(2n)|\theta\right] \\ \\
                                           &= \max_{\theta \leq \theta_0}P\left[2\theta \sum_{i=1}^{n} x_{i}^2 \leq \frac{\theta}{\theta_0}\chi^2_{1-\alpha}(2n)|\theta\right]\\ \\
                                           &=  \max_{\theta \leq \theta_0}P\left[\chi^2_{2n}\leq \frac{\theta}{\theta_0}\chi^2_{1-\alpha}(2n)\right]\end{align*}$$

### (c)

검정법의 power를 구하라

`1` $\beta(\theta) = P\left [2\theta_0\sum x_i^2 \leq \chi^2_{1-\alpha}(2n)\,|\, \theta = \theta_0 \right]$

`2` 귀무가설 하에서 최대값 $\to$ 즉, 제 1종 오류의 최대값

$$\begin{align*} \text{Max}_{\theta \leq \theta_0}\beta(\theta) &= \text{Max}_{\theta \leq \theta_0}P\left [2\theta_0\sum x_i^2 \leq \chi^2_{1-\alpha}(2n)\,|\, \theta = \theta_0 \right]\\ \\ 
                &=\text{Max}_{\theta \leq \theta_0}P\left [2\theta\sum x_i^2 \leq \frac{\theta}{\theta_0}\chi^2_{1-\alpha}(2n)\,|\, \theta = \theta_0 \right] \\ \\ &= \text{Max}_{\theta \leq \theta_0} P \left [ \chi^2(2n) \leq \frac{\theta}{\theta_0}\chi^2_{1-\alpha}(2n)\,|\,\theta_0 \right] \end{align*}$$

`3` 우변의 경우 $\theta$에 대한 증가함수이므로, $\theta = \theta_0$일 때 최대가 된다.!

`4` 따라서 기각역 `R`을 기각역으로 하는 test는

$$H_0: \theta \leq \theta_0 \quad \text{vs} \quad H_1: \theta > \theta_0$$

- 위 가설을 만족하는 모든 $\theta$에 대해서 전부 `Most Powerful test`이므로,  `Uniformly Most Powerful test`이며, size $\alpha$가 된다.

# EX 2

Suppose that $X_1, X_2, . . . , X_n$ is an iid sample from 

$$f_{X}(x|\theta) = \frac{a_0}{\theta^{a_0}} x^{a_0 -1} \,\,I(0<x<\theta)$$

where $a_0 \geq 1$  is a known constant and $\theta > 0$ is an unknown parameter.

(a) Show that $Q = X_{(n)}/\theta$ is a pivotal quantity, where $X_{(n)}$ is the maximum order statistic. Use $Q$ to construct a $1 − \alpha$ confidence set for $\theta$

(b)  Derive a $1 − \alpha$ confidence set for $\theta$ by using the cumulative distribution function of $T = T(X) = X_{(n)}$ as a pivot.

(c) Assume that has an inverse gamma prior distribution with parameters $c > 0$ and $d > 0$ (both known). Write a $1 − \alpha$ credible set for $\theta$.

## sol

### (a)

`사전지식`: pivotal quantity $\to$ 랜덤샘플의 함수이면서, 분포는 $\theta$에 의존하지 않은 성질을 가짐

`1` $Q$의 cdf: $F_Q(q) = P\left(\frac{X_{(n)}}{\theta} \le q\right) = q^{n a_0}$ ($0<q<1$). 모수 $\theta$가 없으므로 축량(Pivot)임.   

$$\begin{align*} F_{Q}(q) &= P\left (\frac{X_{(n)}}{\theta} \leq q\right) \\ \\
                          &= \left [ P(X_{1} \leq \theta q\right ]^n\end{align*}$$

$$\begin{align*}P(X_{1} \leq \theta q) &= \int_{0}^{\theta q} \frac{a_0}{\theta^{a_0}} x_{1}^{a_0 -1}\, dx_1\\ \\ 
                                       &= q^{a_0}\end{align*}$$

$$\divideontimes \,\, F_{Q}(q) =q^{a_0n}\to f_{Q}(q) = na_0q^{na_0-1}$$

- $\theta$와 무관함

`2` 확률식 세우기

$$P(c_1 \le Q = \frac{X_{(n)}}{\theta} \le c_2) = 1 - \alpha$$

`3` 확률식 변형

$$P\left(\frac{X_{(n)}}{c_2}\leq \theta \leq \frac{X_{(n)}}{c_1}\right) = 1-\alpha$$

`4` 따리사, confidence interval set은?>

$$\left[\frac{X_{(n)}}{c_2},\,\,\frac{X_{(n)}}{c_1}\right]$$

---

### (b)

(a)번과 같은 이야기

### (c)

`1` 가능도함수 $L(\theta|\mathbf{x}) \propto \theta^{-n a_0} I(\theta > X_{(n)})$와 사전분포 $\pi(\theta) \propto \theta^{-(c+1)} e^{-d/\theta}$ 곱함. 

`2` 사후분포: $\pi(\theta|\mathbf{x}) \propto \theta^{-(c + n a_0 + 1)} e^{-d/\theta} I(\theta > X_{(n)})$ (즉, 좌측 절단된 역감마 분포).

`3` 균등 꼬리(Equal-tail) 신용구간 $[k_1, k_2]$는 사후 pdf의 적분 조건식으로 정의됨.

`4` 아래 조건을 만족하는 구간 $[k_1, k_2]$

$$\frac{\Gamma\left(c+n a_0, \frac{d}{k_2}\right) - \Gamma\left(c+n a_0, \frac{d}{k_1}\right)}{\gamma\left(c+n a_0, \frac{d}{X_{(n)}}\right)} = 1 - \alpha \quad (\text{단, } X_{(n)} \le k_1 < k_2)$$

(여기서 $\gamma$는 Lower, $\Gamma$는 Upper 불완전 감마 함수)

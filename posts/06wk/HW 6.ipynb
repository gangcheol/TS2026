# ex1

1. Suppose $X_1, X_2, \ldots, X_n$ are iid $N(\mu,\sigma^2)$, where $-\infty < \mu < \infty$ and $\sigma^2 > 0$ and both parameters are unknown. Set $\mathbf{X} = (X_1, X_2, \ldots, X_n)$ and $\theta = (\mu,\sigma^2)$.

   (a) Show that
   $$
   T = T(\mathbf{X}) =
   \begin{pmatrix}
   \overline{X} \\
   S^2
   \end{pmatrix}
   $$
   is a minimal sufficient statistic for $\theta$.

   (b) Define
   $$
   W = \left(\frac{\overline{X}-\mu}{\sigma/\sqrt{n}}\right)^2 + \frac{(n-1)S^2}{\sigma^2}.
   $$
   Find the finite-sample distribution of $W$.

   (c) Is $W$ an ancillary statistic? Explain.

---

## sol

### (a)

`step1`

$$f(\mathbf{x}| \mu, \sigma^2) = (2\pi \sigma^2)^{-n/2} \exp\left\{ -\frac{1}{2\sigma^2} \sum_{i=1}^{n} (x_i-\mu)^2 \right \}$$

`step2`

- 지수부 전개

$$f(\mathbf{x}| \mu, \sigma^2) = (2\pi \sigma^2)^{-n/2} \exp \left ( -\frac{1}{2\sigma^2} \sum_{i=1}^{n} x_i^2\right ) \cdot
\exp \left( \frac{\mu}{\sigma^2} \sum_{i=1}^{n} x_i ^2 - \frac{n\mu^2}{2\sigma^2}\right)$$

즉, 표본은 $\sum x_i, \sum x_i^2$를 통해서만 $(\mu, \sigma^2)$에 의존한다.

그러므로 $U(\mathbf{X}) = \left( \sum_{i=1} X_{i}, \sum_{i=1}^n X_i^2\right)$은 `factorization theorem`에 의해 충분통계량이다.

또한, 아래와 같이 일대일 대응이므로 $(\bar X, S^2)$도 충분통계량이다.

$$\sum_{i=1}^{n} = n\bar X, \quad \sum_{i=1}^{n} X_i^2 = \sum_{i=1}^n (X_i - \bar X)^2 + n\bar X^2 = (n-1)S^{2} + n\bar X^2$$

`step3` Minimal sufficiency

$$\frac{f(\mathbf{x}|\mu, \sigma^2)}{f(\mathbf{y}|\mu, \sigma^2)} = \exp \left [\frac{\mu}{\sigma^2} (\sum x_i - \sum y_i) - 
\frac{1}{2\sigma^2} \sum(x_i^2 - \sum y_i^2)\right ]$$

위 비가 $(\mu,\sigma^2)$의 함수로서 상수가 되려면(즉, $\mu, \sigma^2$에 의존하지 않으려면)

$$\sum x_i = \sum y_i, \quad \sum x_i^2 = \sum y_i^2$$

가 필요충분조건이다. 따라서 $U(\mathbf{\sum X_i, \sum X_i^2 })$는 최소충분성을 같는 충분통계량이고, 문제에서 질문한

$$ T = T(\mathbf{X}) =
   \begin{pmatrix}
   \overline{X} \\
   S^2
   \end{pmatrix} $$

역시 최소충분성을 만족하는 `충분통계량`이다.

---

### (b), (c)

  $$
   W = \left(\frac{\overline{X}-\mu}{\sigma/\sqrt{n}}\right)^2 + \frac{(n-1)S^2}{\sigma^2}.
   $$

위 식을 살펴보면, 왼쪽 항은 표준정규분포(Z)의 제곱항 즉, 자유도가 1인 카이제곱분포를 따르고, 오른쪽 항은 자유도가 $n-1$인 카이제곱분포를 따름을 알 수 있다.

또한 왼쪽과 오른쪽 항은 서로 독립인 $X_i$들의 함수꼴 형태이므로, 서로 독립이기 때문에 `카이제곱분포의 가법성` 성질을 적용하면, 아래와 같이 표현할 수 있다.

$$W \sim \chi_{n}^2$$

위는 주어진 $W$가 모수 $\theta = (\mu, \sigma^2)$에 의존하지 않으므로 $W$는 `Ancillary Statistic`이다.(강의 질문 필요)

## ex2

2. Recall that if $X$ has a $\text{Beta}(\alpha,\beta)$ distribution, then the probability density function (pdf) of $X$ is
   $$
   f_X(x \mid \alpha,\beta)
   = \frac{\Gamma(\alpha+\beta)}{\Gamma(\alpha)\Gamma(\beta)}
   x^{\alpha-1}(1-x)^{\beta-1} I(0<x<1),
   $$
   where $\alpha > 0$ and $\beta > 0$. In this problem, we are going to consider the beta subfamily where $\alpha=\beta=\theta$. Let $X_1, X_2, \ldots, X_n$ denote an iid sample from a beta$(\theta,\theta)$ distribution.

   (a) Show that
   $$
   T(\mathbf{X}) = \prod_{i=1}^n X_i(1-X_i)
   $$
   is a sufficient statistic for $\theta$. Is $T(\mathbf{X})$ complete?

   (b) The two-dimensional statistic
   $$
   T^* = T^*(\mathbf{X}) =
   \begin{pmatrix}
   \prod_{i=1}^n X_i \\
   \prod_{i=1}^n (1-X_i)
   \end{pmatrix}
   $$
   is also a sufficient statistic for $\theta$. What must be true about the conditional distribution
   $$
   f_{\mathbf{X}\mid T^*}(\mathbf{x}\mid \mathbf{t})?
   $$

   (c) Show that $T^*(\mathbf{X})$ is not a complete statistic.

---

## sol

### (a)

`step1`

$$f(\mathbf{x}|\theta) = \prod_{i=1}^{n} \frac{\Gamma(2\theta)}{\Gamma(\theta)^2} x_{i}^{\theta-1}(1-x_i)^{\theta-1} = \left ( \frac{\Gamma(2\theta)}{\Gamma(\theta)^2} \right )^n\prod_{i=1}^{n}  \left (x_{i}(1-x_i)\right)^{\theta-1} = \left ( \frac{\Gamma(2\theta)}{\Gamma(\theta)^2} \right )^n\left (T(X) \right ) ^{\theta-1}$$

`step2`

따라서

$$f(\mathbf{x}| \theta) = g(T(\mathbf{x}), \theta) \cdot h(\mathbf{x})\quad \to \quad g(T(\mathbf{x}), \theta) = \left ( \frac{\Gamma(2\theta)}{\Gamma(\theta)^2} \right )^n\left (T(X) \right ) ^{\theta-1}, \,\,h(x) = 1$$

$T(\mathbf{X})$는 충분통계량이다.

---

### (b)

`step1`

$$f(\mathbf{x}|t)\text{는 모수 }\theta\text{에 의존하면 안됨.}$$

그 분포가 $\theta$가 무엇이든 똑같아야 하기 때문!

즉, $T^{*}$가 충분통계량이면 '$T^{*}$를 조건으로 하는 $\mathbf{X}$의 분포는 $\theta$의 `영향을 받으면 안된다`!'

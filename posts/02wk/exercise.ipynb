## sol

### (a)

$$\begin {align*} \int_{1}^{\infty} e^{-x} 
\left \{ \int_{0}^{1} e^{-2y}\,\,dy\right \} dx &= (1-e^{-2}) [e^{-x}]_{1}^{\infty} \\ \\ 
                                                &= e^{-1}(1-e^{-2})\end{align*}$$

### (b)

$$\begin{align*}\int_{0}^{\infty} 2e^{-2y} 
\left [ \int_{0}^{y} e^{-x}dx\right ]dy &= \int_{0}^{\infty}2e^{-2y}(1-e^{-y}) dy \\ \\
     &= 1 \,+\, \frac {2}{3}(-1) = \frac {1}{3}\end{align*}$$

### (c)

$$f(x) = \int_{y \in (0,\infty)} f(x,y) dy$$

$$\therefore \,\, f(x) = e^{-x}$$

$$ P\{X<a\} =  \int_{0}^{a} f(x)\, dx =  1-e^{-a}$$

$$\begin {align*} f(z) &= \int_{0}^{\infty} f(z,u) \,\,du \\ \\

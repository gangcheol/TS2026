\left [ \int_{0}^{y} e^{-x}dx\right ]dy &= \int_{0}^{\infty}2e^{-2y}(1-e^{-y}) dy \\ \\

# example on page 25 

find the joint probablity mass function of $(X_1, X_2)$

<center><img src = '3.jpg' width = 500></center>

## sol

$X_{1} = \omega_{1} + \omega_{2} = s$

$X_{2} = |\omega_{1} - \omega_{2}| = d$

$\therefore P_{X_{1},X_{2}}(s,d) = P(X_{1} = s, X_{2} = d)$를 구하자.

`1` 조건 정의

- $\omega_1 \geq \omega_2$라고 조건 가정.

$$\omega_1 = \frac{s+d}{2},\quad \omega_2 = \frac{s-d}{2}$$

`2` 따라서 해가 존재하려면, 다음 조건을 만족해야함

- $s \pm d$가 짝수여야함.

-  $1 \leq \omega_1 \leq \omega_2 \leq 6$

-  $d = 0 $: 기능한 개수 $N(s,0) = 1$

-  $d > 0 $: 기능한 개수 $N(s,d) = 2$



`3` 따라서... 최종 joint probablity mass function of $(X_1, X_2)$

$$P_{X_1, X_2}(s,d) = \begin{cases} \frac{1}{36}, \quad d = 0, s \in \{2,4,6,8,10,12\}  \\ 
                               \frac{2}{36}= \frac{1}{18}, \quad d \in \{1,2,3,4,5\}\,\,\text{even},\,\,1 \leq \omega_1 \leq \omega_2 \leq 6\ \\ 
                               0, \quad \text{otherwise} \end{cases}$$

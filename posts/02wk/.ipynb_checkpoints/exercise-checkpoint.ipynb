# example on page 25 

find the joint probablity mass function of $(X_1, X_2)$

<center><img src = '3.jpg' width = 500></center>

## sol

$X_{1} = \omega_{1} + \omega_{2} = s$

$X_{2} = |\omega_{1} - \omega_{2}| = d$

$\therefore P_{X_{1},X_{2}}(s,d) = P(X_{1} = s, X_{2} = d)$를 구하자.

`1` 조건 정의

- $\omega_1 \geq \omega_2$라고 조건 가정.

$$\omega_1 = \frac{s+d}{2},\quad \omega_2 = \frac{s-d}{2}$$

`2` 따라서 해가 존재하려면, 다음 조건을 만족해야함

- $s \pm d$가 짝수여야함.

-  $1 \leq \omega_1 \leq \omega_2 \leq 6$

-  $d = 0 $: 기능한 개수 $N(s,0) = 1$

-  $d > 0 $: 기능한 개수 $N(s,d) = 2$



`3` 따라서... 최종 joint probablity mass function of $(X_1, X_2)$

$$P_{X_1, X_2}(s,d) = \begin{cases} \frac{1}{36}, \quad d = 0, s \in \{2,4,6,8,10,12\}  \\ 
                               \frac{2}{36}= \frac{1}{18}, \quad d \in \{1,2,3,4,5\}\,\,\text{even},\,\,1 \leq \omega_1 \leq \omega_2 \leq 6\ \\ 
                               0, \quad \text{otherwise} \end{cases}$$

# Find the mean and variance of possion by using mgf

## sol

`1` $ X \sim \text{Possion}(\lambda)$

$$f_{X}(x) = \frac{e^{-\lambda}\lambda^{x}}{ x!}$$

`2` mgf

$$M_{X}(t) = E(e^{tX}) = \sum_{x=0}^{\infty}e^{tx} f_{X}(x)$$

$$\therefore \,\, M_{X}(t) = \text{exp}(\lambda(e^{t}-1))$$

`3` mean

$$M^{\prime}_{X}(0) = \lambda$$

`4` variance

$$E(X^{2}) = M_{X}^{\prime\prime}(0) = M_{X}(0)(\lambda + \lambda^2) =(\lambda + \lambda^2)$$

$$\begin{align*}Var(X) &= E(X^{2}) - [E(X)]^{2} \\ \\
                       &= (\lambda + \lambda^2) - \lambda^2 \\ \\
                       &= \lambda \end{align*}$$

## PoI approximation to bin

$X \sim \text{Bin}(100,000, 0.00002), \text{find} \quad P(10<X<10)$ using poisson approximation to binomial distribution.

## sol

`step1`

n이 매우 크고, p가 매우 작으며, $\lambda = np$가 적잗한 크기일 떄, 포아송 근사를 만족

$$n = 100,000, \,\, p = 0.00002$$

$$\lambda = np = 2$$

`step2`

$$10 < X < 20 \to X \in \{11,12\dots 19\}$$

`step3`

$Y \sim \text{POI}(2)$

$$P(Y = k) = e^{-2} \frac{2^{k}}{k!}$$

`step4`

$$P(10<X<20)  \approx \sum_{k=11}^{19}\frac{2^{k}}{k!} \approx 8.31 \times 10^{-6}$$

# Suppose1...

Suppose that $X \sim \text{bin}(n,p), \,\, Y \sim \text{bin}(m,p),\, X\text {and} Y$ are indep, Find

$$P\{X = i| X+Y=k\}, \quad 0<i<k<n.$$

## sol

`step1` $X,Y$는 독립이므로 두 확률변수의 결합확률밀도함수를 다음과 같이 쓸 수 있음.

$$P\{X = i, X+Y=k\} = P\{X = i, Y=k-i\} = P(X = i)P(Y = k-i).$$

`step2`

$$P(X=i) = \binom{n}{i}p^{i}(1-p)^{n-i}, \quad P(Y=k-i)=\binom{m}{k-i}p^{k-i}(1-p)^{m-(k-i)}$$

`step3`

$$P(X = i, Y=k-i)=\binom{n}{i}\binom{m}{k-i}p^{k}(1-p)^{n+m-k}$$

`step4` 두 확률변수는 같은 $p$를 갖는 독립 이항분포를 따르므로

$$X + Y \sim \text{bin}(n+m,p)$$

$$\therefore\,\,  P(X=i|X+Y=k) = \frac{\binom{n}{i}\binom{m}{k-i}}{\binom{n+m}{k}} $$

# Suppose2...

Suppose $Y_{1}\dots Y_{n}$ is random variable sample from the population with distribution $\text{Gamma}(1,2)$. Find the distribution of min$(Y_{1}\dots Y_n)$

## sol

`step1`

$$Y_{i} \sim \text{Gamma}(\alpha = 1, \theta = 2) $$

$$f_{Y}(y) = \frac{1}{2} e^{-y/2}$$

`step2`

$$F_{M}(y) = P(\text{min}(Y_1 \dots Y_n) \leq y)$$ 

`step3`

$$F_{M}(y) = 1- P(Y_{1}>y, \dots Y_{n}>y)$$ 

`step3` 독립성 사용

$$P(Y_{1}>y, \dots Y_{n}>y) = \prod_{i=1}^{n} P(Y_{i}>y) = \left ( P(Y_1 > y) \right )^{n}$$

`step4`

$$ P(Y_1 > y) = 1- P(Y_{1}<y)= 1- F_{Y}(y)$$

$$F_{Y}(y) = \int_{0}^{y} \frac{1}{2}e^{-t/2}\,dt = 1-e^{-y/2}$$

$$ P(Y_1 > y) = e^{-y/2}$$

`step5` 최종 CDF

$$F_{M}(y) = 1- (e^{-y/2})^{n} = 1- e^{-ny/2}$$

`step6` 최종 pdf: $y$에 대해서 미분

$$f_{M}(y) = \frac{n}{2} e^{-ny/2}$$

# Suppose3....

$Y_1\dots Y_n$ is a random sample from the population with distribution $N(\mu,\sigma^{2})$. Find followings.

## (1) The distribution of Y1 + 2Y2 + 3Y3

### sol

세 확률변수는 독립이고, 독립인 확률변수의 분포는 선형결합을 따르므로

- 독립이므로 공분산은 `0`

$$Y_{1} + 2Y_{2} + 3Y_{3} \sim N(6\mu, 14\sigma^2) $$

## (2) The variance of S2

$$S^{2} = \sum_{i=1}^{n-1} \frac{(Y_{i}-\bar{Y})^2}{n-1}$$

### sol

`step1`

$$W=\frac{(n-1)S^{2}}{ \sigma^2} \sim \chi^{2}_{n-1}$$

$$\therefore\,\, S^{2}  = \frac{\sigma^2}{n-1}W$$

결론

$$Var(S^{2}) = \left ( \frac{\sigma^2}{n-1}\right)^2 Var(W) = \frac {2\sigma^4}{n-1}$$

## (3) The distribution of

$Y_2 - \bar Y$ and its mean and variance

### sol.

각 확률변수는 독립이므로 선형결합인 $\bar Y$와 $Y_2$도 독립

$E(Y_2) = \mu, E(\bar Y) = \mu$

$Var(Y_2) = \sigma ^2, Var(\bar Y) = \frac{\sigma^2}{n}$

$$Cov(Y_2, \bar Y) = \frac{1}{n} \sum_{i=1}^{n}Cov(Y_2, Y_i) = \frac 1n Cov(Y_2,Y_2) = \frac 1n\sigma^2$$

$$Var(Y2-\bar Y) = Var(y_2) - Var(\bar Y) -2Cov(Y_2, \bar Y) = \frac{n-1}{n}\sigma^2$$

$$\therefore\,\, Y_2 - \bar y \sim N(0, \frac{n-1}{n}\sigma^2)$$

## (4) Cov...

$$Cov(\bar Y, Y_2-\bar Y)$$

### sol.

`step1` 공분산의 선형성

$$Cov(\bar Y, Y_2-\bar Y) = Cov(\bar Y, Y_2) - Cov(\bar Y, \bar Y) = Cov(\bar Y, Y_2) - Var(\bar Y)$$

`step2`

$$Cov(\bar Y, Y_2) = \frac{\sigma^2}{n}, \quad Var(\bar Y) = \frac{\sigma^2}{n}$$

$$\therefore\,\, Cov(\bar Y, Y_2-\bar Y) =  \frac{\sigma^2}{n} -  \frac{\sigma^2}{n} = 0$$

## (5) independent?

Are $Y_1 - Y_2$ and $Y_1 + Y_2$ independent?

### sol.

$$U = Y_1+ Y_2,\,\,V = Y_1-Y_2$$

`1` 정규분포를 따르는 두 확률변수의 선형변환은 다시 정규분포를 따르므로 위 $(U,V)$는 이변량 정규분포를 따른다.

`2` 공분산 계산

$$Cov(U,V) = Cov(Y_1, Y_1) - Cov(Y_1,Y_2) + Cov(Y_2,Y_1) - Cov (Y_2,Y_2) = 0$$

`3` 결론

$(U,V)$가 이변량(jointly)정규분포를 따르고, $Cov(U,V)$가 0 이므로

$$U,V \text{는 서로 독립이다.}$$

## (6) The distribution of

$$\sum_{k=1}^{5} \frac{(Y_{2k-1}-Y_{2k})^2}{2\sigma^2}$$

### sol

`step1`

$$D_{k} = Y_{2k-1} - Y_{2k} \sim N(0, 2\sigma^2)$$

`step2`

$$Z_{k} = \frac{D_{k}}{\sqrt {2}\sigma} \sim N(0,1)$$

$$Z_{k}^2 \sim \chi_{1}^{2}$$

`step3`

서로 독립인 카이제곱분포를 따르는 확률변수들의 가법성 성질을 이용하면!

$$\therefore \,\,T = \sum_{k=1}^{5} Z_{k}^2 = \sum_{k=1}^{5} \frac{(Y_{2k-1}-Y_{2k})^2}{2\sigma^2} \sim \chi^{2}_{5}$$

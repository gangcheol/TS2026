# EX 1

Suppose $X_1, X_2, \dots X_n$ are iid sample from $f_{X}(x|\theta)$ where $\theta \in \Theta \subset \mathbb{R}$. Recall the score function is given by

$$S(\theta|x) = \frac{\partial}{\partial} \log{f_{X}(x|\theta)}$$

where $f_{X}(x|\theta)$ is the joint distribution of $X = (X_{1},X_{2}, \dots X_{n})$

## (a)

Show that $E_{\theta}[S(\theta|X)] = 0$

### sol

$$\begin {align*}E_{\theta}[S(\theta|X)]  &= \int S(\theta|x) f_{X}(x|\theta) \,dx \\ \\ 
                                          &= \int \left ( \frac{\partial}{\partial} \log{f_{X}(x|\theta)} \right ) f_{X}(x|\theta)\,dx \\ \\
                                          &= \int \left ( \frac{\frac{\partial}{\partial}f_{X}(x|\theta) }{f_{X}(x|\theta)} \right ) f_{X}(x|\theta)\,dx \\ \\ 
                                          &= \frac{\partial}{\partial \theta}1 = 0\end{align*}$$

$$S(\theta|x) = \sum_{i=1}^n  \frac{2(x_i-\theta)}{1 + (x_i - \theta)^2}$$

# EX 2

Suppose $X_1, \dots X_n$ are iid $N(\theta, 1)$, where $-\infty < \theta < \infty.$

## (a)

Show that $W(X) = \bar X$  is the uniformly minimum variance unbiased estimator (UMVUE) for $\theta$ in two ways: one way that uses a Cram´er-Rao Lower Bound argument and one
way that uses sufficiency and completeness.

### sol

`1` 로그-확률밀도함수 편미분

$$\frac{\partial}{\partial \theta} \log{f(x_i|\theta)} = x_i - \theta \quad \to \quad \frac{\partial^{2}}{\partial \theta^2} \log {f(x_i|\theta)} = - 1$$

`2` 피셔 정보량 계산

$$I_1(\theta) = -E\left [\frac{\partial^{2}}{\partial \theta^2} \log {f(x_i|\theta)}\right] = -E[-1] = 1$$

$$\divideontimes \quad I_{n}(\theta) = n$$

`3`  $\frac{1}{I_{n}(\theta)}= \frac{1}{n}$

`4` $\overline {X}$의 분산 $\to$ $Var{(\overline{X})} = \frac{1}{n}$

따라서 불편추정량 $\bar X$의 분산이 크레모-라오 하한값($1/n$)에 수렴하므로, $\overline X$는 `UMVUE`이다.

---

## (b)

Find the `UMVUE` for $\tau(\theta) = e^{\theta}.$

### sol

`1` $\bar x \sim N(\theta, \frac 1n)$

`2` $e^{\theta}$의 기댓값 계산

$$E[e^{\overline{X}}] = \exp\left( \theta \cdot 1 + \frac{1}{2} \cdot \frac{1}{n} \cdot 1^2 \right) = e^\theta \cdot e^{\frac{1}{2n}}$$

`3` 불편성 확인

$$E\left[ \frac{e^{\overline{X}}}{e^{\frac{1}{2n}}} \right] = e^\theta \implies E\left[ e^{\overline{X} - \frac{1}{2n}} \right] = e^\theta$$

`4` 결론

$e^{\overline{X} - \frac{1}{2n}}$ 는 $\tau(\theta)$의 `MVUE`이다.

## (c)

Calculate $E(X1|X), E(X1 − X|X)$, and $Cov(X1, X1 − X|X)$.

### sol

`1` $E(X_1|\overline X)$ 계산

$$\sum_{i=1}^{n} E(X_i|\overline{X}) = E\left(\sum_{i=1}^{n} X_i \;\middle|\; \overline{X}\right) = E(n\overline{X}|\overline{X}) = n\overline{X}$$

$$n \cdot E(X_1|\overline{X}) = n\overline{X} \implies \mathbf{E(X_{1}|\overline{X}) = \overline{X}}$$

`2` $E(X_{1}-\overline{X}|\overline{X})$ 계산

$$E(X_{1}-\overline{X}|\overline{X}) = E(X_1|\overline{X}) - E(\overline{X}|\overline{X})$$

$E(X_1|\overline{X}) = \overline{X}$이므로

$$= \overline{X} - \overline{X} = \mathbf{0}$$

`3`  $\text{Cov}(X_{1}, X_{1}-\overline{X}|\overline{X})$ 계산

$$\text{Cov}(X_1 - \overline{X}, \overline{X}) = \text{Cov}(X_1, \overline{X}) - \text{Var}(\overline{X}) = \frac{1}{n} - \frac{1}{n} = 0$$

$X_1 - \overline{X}$와 $\overline{X}$는 서로 독립(Independent), 따라서...

$$\text{Cov}(X_{1}, X_{1}-\overline{X}|\overline{X}) = \text{Cov}(X_1, X_1 - \overline{X}) = \text{Var}(X_1) - \text{Cov}(X_1, \overline{X}) = \mathbf{1 - \frac{1}{n}}$$

# EX 3

Suppose $X_1, X_2 \dots X_m$ are iid $N(0, \sigma_1^2)$. Suppose $Y_1, Y_2 \dots Y_n$ are iid $N(0, \sigma_2^2)$. Suppose the samples are independent.

## (a)

Derive the likelihood ratio test (LRT) statistic $\lambda (x, y)$ for testing

$$H_0: \sigma_1^2 = \sigma_2^{2} \quad vs \quad H_1: \text{not } H_0$$

and show that it is a function of $t_1 = t(x) = \sum_{i=1}^{m} x_i^{2}$ and $t_2 = t_2(y) = \sum_{j=1}^{n} y_{j}^{2}$

### sol

`1` 결합가능도 함수

$$L(\sigma_1^2, \sigma_2^2) = (2\pi)^{-\frac{m+n}{2}} (\sigma_1^2)^{-\frac{m}{2}} (\sigma_2^2)^{-\frac{n}{2}} \exp\left( -\frac{t_1}{2\sigma_1^2} - \frac{t_2}{2\sigma_2^2} \right)$$

`2` 대립가설 하에서의 최대 가능도

$$\hat{\sigma}_1^2 = \frac{t_1}{m}, \quad \hat{\sigma}_2^2 = \frac{t_2}{n}$$

$$L(\hat{\Omega}) = (2\pi)^{-\frac{m+n}{2}} \left(\frac{t_1}{m}\right)^{-\frac{m}{2}} \left(\frac{t_2}{n}\right)^{-\frac{n}{2}} e^{-\frac{m+n}{2}}$$

`3` 귀무가설 하에서의 최대 가능도

$$L(\hat{\Omega}_0) = (2\pi)^{-\frac{m+n}{2}} \left(\frac{t_1+t_2}{m+n}\right)^{-\frac{m+n}{2}} e^{-\frac{m+n}{2}}$$

`4` LRT  통계량

$$\lambda(x,y) = \frac{L(\hat{\Omega}_0)}{L(\hat{\Omega})} = \frac{\left(\frac{t_1+t_2}{m+n}\right)^{-\frac{m+n}{2}}}{\left(\frac{t_1}{m}\right)^{-\frac{m}{2}} \left(\frac{t_2}{n}\right)^{-\frac{n}{2}}} = \frac{\left(\frac{t_1}{m}\right)^{\frac{m}{2}} \left(\frac{t_2}{n}\right)^{\frac{n}{2}}}{\left(\frac{t_1+t_2}{m+n}\right)^{\frac{m+n}{2}}}$$

`5` 결론

$\lambda (x,y)$는 오직 $t_1, t_2$만의 함수로 표현됨.

---

## (b)

Show how you could perform a size test in part (a) using the $F$ distribution

### sol

`1` F분포 정의

$$F = \frac{t_1 / m}{t_2 / n} \sim F_{m, n}$$

`2` F에 관한 식으로 정리

$$\lambda = \text{상수} \times \frac{F^{m/2}}{\left(\frac{m}{n}F + 1\right)^{(m+n)/2}}$$

`3` 유의수준 $\alpha = 0.05$ 검정 수행

$$\text{Rejection Region: } F < F_{0.975, m, n} \quad \text{또는} \quad F > F_{0.025, m, n}$$

$$F_{0.975, m, n} = \frac{1}{F_{0.025, n, m}}$$

`4` 결론

검정통계량 $F$의 값이 기각역에 속하면, 즉 $F < F_{0.975, m, n}$ 이거나 $F > F_{0.025, m, n}$ 이면 유의수준 0.05에서 귀무가설 $H_0$를 기각하고 두 집단의 분산이 통계적으로 유의미하게 다르다고 결론을 내릴 수 있다.

# 4번

Suppose $X_1$ and $X_2$ are iid from

$$f_{X}(x|\theta) = \theta x^{\theta - 1} I(0<x<1)$$

where $\theta >0$, and consider testing $H_0 : \theta \leq 1$ versus $H_1: \theta > 1$. We have two tests:

$$\phi_1 = \phi_2(x_1, x_2) = I(x_1 > 0.9)$$

$$\phi_2 = \phi_2(x_1, x_2) = I(x_1x_2 > c)$$

where $0<c<1$.

## (a)

Show that the power functions of the two tests are $\beta_{1}(\theta) = 1- (0.9)^{\theta}$ and $\beta_{2}(\theta) = 1 + c^{\theta}(\theta \log c -1)$, respectively

### sol

`1` $\beta_1$

$$\beta_1(\theta) = P_\theta(X_1 > 0.9) = \int_{0.9}^{1} \theta x^{\theta-1} dx = \left[ x^\theta \right]_{0.9}^{1} = \mathbf{1 - (0.9)^\theta}$$

`1` $\beta_2$

$$\beta_2(\theta) = P_\theta(X_1 X_2 > c) = \int_{c}^{1} \int_{c/x_1}^{1} \theta^2 x_1^{\theta-1} x_2^{\theta-1} dx_2 dx_1$$

$$= \int_{c}^{1} \theta x_1^{\theta-1} \left[ x_2^\theta \right]_{c/x_1}^{1} dx_1 = \int_{c}^{1} \theta x_1^{\theta-1} \left( 1 - \frac{c^\theta}{x_1^\theta} \right) dx_1$$$$= \int_{c}^{1} \left( \theta x_1^{\theta-1} - \theta c^\theta x_1^{-1} \right) dx_1$$

$$= \left[ x_1^\theta \right]_c^1 - \theta c^\theta \left[ \log x_1 \right]_c^1 = (1 - c^\theta) - \theta c^\theta (0 - \log c)$$$$= \mathbf{1 + c^\theta(\theta\log c - 1)}$$

---

## (b)

Calculate the size of the $\phi_1$ test. Then, find the value of $c$ that gives the same size for the $\phi_2$ test

### sol

`1` $\beta_1(\theta) = 1- (0.9)^{\theta}$ 는 $\theta$에 대해서 단조 증가함수이므로,

$$\text{Size of } \phi_1 = \beta_1(1) = 1 - 0.9 = \mathbf{0.1}$$

`2` $\beta_2(\theta)$도 마찬가지.

$$\text{Size of } \phi_2 = \beta_2(1) = 1 + c(\log c - 1) = 1 + c\log c - c$$

`3` 방정식 세우기

$$1 + c\log c - c = 0.1 \implies c - c\log c = 0.9$$

$$\divideontimes \,\, c \approx 0.588$$

# EX 1

Suppose $X_1, X_2, \dots X_n$ are iid sample from $f_{X}(x|\theta)$ where $\theta \in \Theta \subset \mathbb{R}$. Recall the score function is given by

$$S(\theta|\boldsymbol{x}) = \frac{\partial}{\partial\,\theta} \log{f_{\boldsymbol{X}}(\boldsymbol{x}|\theta)}$$

where $f_{X}(\boldsymbol{x}|\theta)$ is the joint distribution of $X = (X_{1},X_{2}, \dots X_{n})$

## (a)

Show that $E_{\theta}[S(\theta|X)] = 0$

### sol

`step1` score function 전개

$$S(\theta|\boldsymbol{x}) = \sum_{i=1}^{n}\frac{\partial}{\partial\,\theta} \log f_{x_i}(x_i|\theta)$$

`step2` 기대값 공식에 대입

$$\begin{align*}E_{\theta}[ S(\theta|X)] &= \int \sum_{i=1}^{n}\frac{\partial}{\partial\,\theta} \log f_{x_i}(x_i|\theta) \times f_{x_i}(x_{i}|\theta) \,dx\\ \\
                                         &= n \times \int  \frac{f^{\prime}(x|\theta)}{f(x|\theta)} \times f(x|\theta) \,dx \\ \\ 
                                         &= n \times \int f^{\prime}(x|\theta) \, dx \\ \\
                                         & = n\times  \frac{\partial}{\partial\, \theta}\int f(x|\theta) \,dx= n \times 0 = 0 \end{align*}$$

$$\divideontimes \,\, E_{\theta}[S(\theta|\boldsymbol{X}) ] =0\text{ 이다}$$

---

## (b)

Calculate the score function when

$$f(x|\theta) = \frac{1}{\pi} \frac{1}{1+(x-\theta)^2}$$

### sol

`1` joint pdf 구하기

$$L(\theta) = f_{\boldsymbol {X}} (\boldsymbol{x}|\theta) = \prod_{i=1}^{n} \frac{1}{\pi} \frac{1}{1+(x_i-\theta)^2}$$

`2` 로그 변환

$$\begin{align*} l(\theta) &= \log L(\theta)  \\ \\
                           &= -n\log \pi -\sum \log (1+(x_i-\theta)^2)\end{align*}$$

`3` score function에 대입, 즉 위 식을 $\theta$에 대하여 미분

$$\begin{align*} S(\theta|\boldsymbol{x}) &= \frac{\partial}{\partial\,\theta} \log{f_{\boldsymbol{X}}(\boldsymbol{x}|\theta)} \\ \\
                                          &= \sum_{i=1}^{n}\frac{2(x_i-\theta^2)}{1+(x_i-\theta)^2}\end{align*}$$

## (c)

How would you find the maximum likelihood estimate $\theta$?

### sol

score function이 0이 되는 $\theta$값을 구해야하나, 손으로 계산이 안됨(`analytical solution`이 없는 경우임)

# EX 2

Suppose $X_1, \dots X_n$ are iid $N(\theta, 1)$, where $-\infty < \theta < \infty.$

## (a)

Show that $W(X) = \bar X$  is the uniformly minimum variance unbiased estimator (UMVUE) for $\theta$ in two ways: one way that uses a Cram´er-Rao Lower Bound argument and one
way that uses sufficiency and completeness.

### sol 1.  Cramer-Rao Lower Bound

`tip` 어떤 추정량 $T$가 비편향 추정량이고, $Var(T) \geq \frac{1}{nI(\theta)}$, 이면 크레모-라온 부등식에 의해 $T$는 UMVUE임이 자명함.

`step1` 비편향 추정량 확인

$E\,[W(\boldsymbol{X})] = E[\bar{X}]= \theta$ 이므로 비편향 추정량임

`step2` $Var(\bar{X})= \frac{1}{n}$

`step3` log pdf를 두번 미분한 값 구하기

1. $f(x) = (2\pi)^{-1/2}\exp(-(x-\theta)^2/2)$

2. $\log f(x|\theta) = -\frac 12 \log 2\pi - \frac 12 (x-\theta)^2$

3. 로그 pdf 두번 미분

$$\frac{\partial^2}{\partial \theta^2} = 1$$

`step4` 피셔의 정보량 $I(\theta)$ 구하기

- 위에서 구한 식을 아래 공식에 대입

$$\begin{align*} I(\theta) &= E\left [ \frac{\partial^2}{\partial \theta^2} \log f_{X} (X|\theta)\right ] \\ \\
                           &= 1\end{align*}$$

`step5` 비교

$Var(\bar{X})= \frac{1}{n}$, $nI(\theta)= n$ 

이므로,  $Var(T) \geq \frac{1}{nI(\theta)}$을 만족함.

$$\divideontimes \,\, W(\boldsymbol{X}) = \bar X = \hat{\theta}_{UMVUE}$$

---

### sol2. sufficiency and completeness.

`tip`

만약 어떠한 추정량 $W(X)$가 비편향추정량이고, $T$가 완비충분통계량이면,

$$E[W(X)|T] = \hat{\theta}_{UMVUE}$$

`1` pdf 재확인

$$\begin{align*} \\ f(x) &= (2\pi)^{-1/2}\exp(-(x-\theta)^2/2) \\ \\
                         &= (2\pi)^{-1/2}\exp\left [-\frac{1}{2} (x^{2}-2x\theta+\theta^2)\right]\end{align*}$$

`2` 따라서 지수족 정의에 의해

- $T(X)=\sum {X_i} = \text{C.S.S}$ 이다.

`3` 그렇담 $\frac{1}{n} \sum_{X_i}$도 `C.S.S`이고, 비편향추정량이므로

$$\divideontimes \,\, W(\boldsymbol{X}) = \bar X = \hat{\theta}_{UMVUE}$$

---

## (b)

Find the `UMVUE` for $\tau(\theta) = e^{\theta}.$

### sol

`1` $\bar x \sim N(\theta, \frac 1n)$

`2` $e^{\theta}$의 기댓값 계산(적률생성함수 이용)

$$\begin{align*}E[e^{\bar X}] &= M_{\bar X}(1) \\ \\
                              &= \exp (\mu t + \sigma^2t^2/2) \\ \\
                              &= \exp (\theta + 1/2n)\end{align*}$$

`3` 불편성 확인

$$E\left[ \frac{e^{\overline{X}}}{e^{\frac{1}{2n}}} \right] = e^\theta \implies E\left[ e^{\overline{X} - \frac{1}{2n}} \right] = e^\theta$$

`4` 결론

$e^{\overline{X} - \frac{1}{2n}}$ 는 $\tau(\theta)$의 `MVUE`이다.

---

## (c)

Calculate $E(X_1|\overline X), E(X_1 − \overline X|\overline X)$, and $Cov(X_1, X_1 − \overline X|\overline X)$.

### sol

`1` $E(X_1|\overline X)$ 계산

$$\sum_{i=1}^{n} E(X_i|\overline{X}) = E\left(\sum_{i=1}^{n} X_i \;\middle|\; \overline{X}\right) = E(n\overline{X}|\overline{X}) = n\overline{X}$$

$$n \cdot E(X_1|\overline{X}) = n\overline{X} \implies {E(X_{1}|\overline{X}) = \overline{X}}$$

`2` $E(X_{1}-\overline{X}|\overline{X})$ 계산

$$E(X_{1}-\overline{X}|\overline{X}) = E(X_1|\overline{X}) - E(\overline{X}|\overline{X})$$

$E(X_1|\overline{X}) = \overline{X}$이므로

$$= \overline{X} - \overline{X} = \mathbf{0}$$

`3`  $\text{Cov}(X_{1}, X_{1}-\overline{X}|\overline{X})$ 계산

- 들어가기전...

$$Cov(X, Y|Z) = E[(X-E[X|Z]) \,\times\,(Y-E[Y|Z])]$$

- 위를 대입

$$\begin{align*} Cov(X_1, X_1 - \overline X|\overline X) &= E\left[(X_1 - E[X_1|\overline X]) \times ([X_1-\overline{X}]  - E[X_1-\overline{X}|\overline X]) \right] \\ \\ 
                    &= E\left [ (X_1-\overline X)^2|\overline X\right]\end{align*}$$

- $X_1 - \overline X \sim N(0, 1-\frac 1n)$

$$ E\left [ (X_1-\overline X)^2|\overline X\right] = Var(X_1 - \overline {X}) = 1-\frac{1}{n}$$

$$\divideontimes \,\, \text{Cov}(X_1, X_1-\overline {X}|\overline{X}) = 1- \frac 1n$$

# EX 3

Suppose $X_1, X_2 \dots X_m$ are iid $N(0, \sigma_1^2)$. Suppose $Y_1, Y_2 \dots Y_n$ are iid $N(0, \sigma_2^2)$. Suppose the samples are independent.

## (a)

Derive the likelihood ratio test (LRT) statistic $\lambda (x, y)$ for testing

$$H_0: \sigma_1^2 = \sigma_2^{2} \quad vs \quad H_1: \text{not } H_0$$

and show that it is a function of $t_1 = t(x) = \sum_{i=1}^{m} x_i^{2}$ and $t_2 = t_2(y) = \sum_{j=1}^{n} y_{j}^{2}$

### sol

`1` 결합가능도 함수

$$L(\sigma_1^2, \sigma_2^2) = (2\pi)^{-\frac{m+n}{2}} (\sigma_1^2)^{-\frac{m}{2}} (\sigma_2^2)^{-\frac{n}{2}} \exp\left( -\frac{t_1}{2\sigma_1^2} - \frac{t_2}{2\sigma_2^2} \right)$$

`2` 대립가설 하에서의 최대 가능도

$$\hat{\sigma}_1^2 = \frac{t_1}{m}, \quad \hat{\sigma}_2^2 = \frac{t_2}{n}$$

$$L(\hat{\Omega}) = (2\pi)^{-\frac{m+n}{2}} \left(\frac{t_1}{m}\right)^{-\frac{m}{2}} \left(\frac{t_2}{n}\right)^{-\frac{n}{2}} e^{-\frac{m+n}{2}}$$

`3` 귀무가설 하에서의 최대 가능도

$$L(\hat{\Omega}_0) = (2\pi)^{-\frac{m+n}{2}} \left(\frac{t_1+t_2}{m+n}\right)^{-\frac{m+n}{2}} e^{-\frac{m+n}{2}}$$

위를 미분해서 최대가능도를 구하면

$$\hat{\sigma^2} = \frac{t_1+t_2}{m+n}$$

`4` LRT  통계량

$$\lambda(x,y) = \frac{L(\hat{\Omega}_0)}{L(\hat{\Omega})} = \frac{\left(\frac{t_1+t_2}{m+n}\right)^{-\frac{m+n}{2}}}{\left(\frac{t_1}{m}\right)^{-\frac{m}{2}} \left(\frac{t_2}{n}\right)^{-\frac{n}{2}}} = \frac{\left(\frac{t_1}{m}\right)^{\frac{m}{2}} \left(\frac{t_2}{n}\right)^{\frac{n}{2}}}{\left(\frac{t_1+t_2}{m+n}\right)^{\frac{m+n}{2}}}$$

`5` 가설검정

$$P\left(\lambda(x,y) = \frac{L(\hat{\Omega}_0)}{L(\hat{\Omega})} = \frac{\left(\frac{t_1+t_2}{m+n}\right)^{-\frac{m+n}{2}}}{\left(\frac{t_1}{m}\right)^{-\frac{m}{2}} \left(\frac{t_2}{n}\right)^{-\frac{n}{2}}} = \frac{\left(\frac{t_1}{m}\right)^{\frac{m}{2}} \left(\frac{t_2}{n}\right)^{\frac{n}{2}}}{\left(\frac{t_1+t_2}{m+n}\right)^{\frac{m+n}{2}}} \leq C \,|\,\sigma_1^2 = \sigma_2^2\right) = \alpha $$

위를 만족하는 기각역을 구하면 된다.

---

## (b)

Show how you could perform a size test in part (a) using the $F$ distribution

### sol

`1` 카이제곱 분포 변환

$$T_{1} = \sum_{i=1}^{m} \frac{X_i^2}{\sigma_1^2} \sim \chi^2_{m}$$

$$T_{2} = \sum_{j=1}^{n} \frac{Y_i^2}{\sigma_1^2} \sim \chi^2_{n}$$

`3` F분포 기각역 설정

- F분포 정의 복기...

$$R = \frac{\sum Y^2/\sigma^2n}{\sum X^2/\sigma^2m} = F(n,m) \leq C_1$$

`4` 결론: 아래를 만족하는 영역이 유의수준 $\alpha$에서의 기각영역이다.

$$P\left[\frac{\sum Y^2/n}{\sum X^2/m} \leq F_{1-\frac{\alpha}{2}}(n,m) \text{ or } \frac{\sum Y^2/n}{\sum X^2/m} \geq F_{1-\frac{\alpha}{2}}(n,m)\right] = \alpha$$

# 4번

Suppose $X_1$ and $X_2$ are iid from

$$f_{X}(x|\theta) = \theta x^{\theta - 1} I(0<x<1)$$

where $\theta >0$, and consider testing $H_0 : \theta \leq 1$ versus $H_1: \theta > 1$. We have two tests:

$$\phi_1 = \phi_2(x_1, x_2) = I(x_1 > 0.9), \quad \text{기각역 } R_1 $$

$$\phi_2 = \phi_2(x_1, x_2) = I(x_1x_2 > c), \quad \text{기각역 } R_2$$

where $0<c<1$.

## (a)

Show that the power functions of the two tests are $\beta_{1}(\theta) = 1- (0.9)^{\theta}$ and $\beta_{2}(\theta) = 1 + c^{\theta}(\theta \log c -1)$, respectively

### sol

`1` $\beta_1$: 귀무가설을 기각할 확률

$$\beta_1(\theta) = P_\theta(X_1 > 0.9) = \int_{0.9}^{1} \theta x^{\theta-1} dx = \left[ x^\theta \right]_{0.9}^{1} = \mathbf{1 - (0.9)^\theta}$$

`1` $\beta_2$

$$\beta_2(\theta) = P_\theta(X_1 X_2 > c) = \int_{c}^{1} \int_{c/x_1}^{1} \theta^2 x_1^{\theta-1} x_2^{\theta-1} dx_2 dx_1$$

$$= \int_{c}^{1} \theta x_1^{\theta-1} \left[ x_2^\theta \right]_{c/x_1}^{1} dx_1 = \int_{c}^{1} \theta x_1^{\theta-1} \left( 1 - \frac{c^\theta}{x_1^\theta} \right) dx_1$$$$= \int_{c}^{1} \left( \theta x_1^{\theta-1} - \theta c^\theta x_1^{-1} \right) dx_1$$

$$= \left[ x_1^\theta \right]_c^1 - \theta c^\theta \left[ \log x_1 \right]_c^1 = (1 - c^\theta) - \theta c^\theta (0 - \log c)$$$$= \mathbf{1 + c^\theta(\theta\log c - 1)}$$

---

## (b)

Calculate the size of the $\phi_1$ test. Then, find the value of $c$ that gives the same size for the $\phi_2$ test

- 귀무가설 하에서 검정력 함수의 최대값(size)를 구하라는 말임

### sol

`1` $\beta_1(\theta) = 1- (0.9)^{\theta}$ 는 $\theta$에 대해서 단조 증가함수이므로,

$$\text{max}_{H_{0}}(\beta_1|\theta) = \beta_1(1) = 1 - 0.9 = \mathbf{0.1}$$

`2` $\beta_2(\theta)$도 마찬가지.

$$\text{Size of } \phi_2 = \beta_2(1) = 1 + c(\log c - 1) = 1 + c\log c - c$$

`3` 방정식 세우기

$$1 + c\log c - c = 0.1 \implies c - c\log c = 0.9$$

$$\divideontimes \,\, c \approx 0.588$$

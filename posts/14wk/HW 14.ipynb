# EX 1

Suppose $X_{1},X_{2},...,X_{n}$ are iid with continuous cdf F. Define

$$\phi_{p}=F^{-1}(p)=\inf\{x\in\mathbb{R}:F(x)\ge p\}$$.

We call $\phi_{p}$ the pth quantile of the distribution of X. Note that if F is strictly increasing, then $F^{-1}(p)$ is well defined by

$$F(F^{-1}(p)) = p$$

The simplest definition of the sample pth quantile is $\hat{F}_{n}^{-1}(p)$ , where

is the empirical distribution function (edf). $$\hat{F}_{n}(x)=\frac{1}{n}\sum_{i=1}^{n}I(X_{i}\le x)$$

is the empirical distribution function (edf). The edf is a non-decreasing step function that takes steps of size $1/n$ at each observed $X_i$. Therefore,

$$\hat{\phi}_p \equiv \hat{F}_n^{-1}(p) = \begin{cases} X_{(np)}, & np \in \mathbb{Z}^+ \\ X_{([np]+1)}, & \text{otherwise} \end{cases}$$

(a) Prove that as $n\rightarrow\infty$,

$$\sqrt{n}(\hat{\phi}_{p}-\phi_{p})\xrightarrow{d}\mathcal{N}\left(0,\frac{p(1-p)}{f^{2}(\phi_{p})}\right)$$

 where f is the population pdf of X.  

(b  If $p=0.5$, then $\phi_{p}=\phi_{0.5}$ is the median of X. Suppose $X_{1},X_{2},...,X_{n}$ are iid $\mathcal{N}(\mu,\sigma^{2})$, where both parameters are unknown. Consider the following two estimators $W_{n}=\overline{X}_{n}$ and $V_{n}=\hat{\phi}_{0.5}$. Find the distribution of $W_{n}$, the limiting distribution of $V_{n}$ and the $ARE(V_{n}\text{ to }W_{n})$. 

## sol

### (a)

`1` 임의의 실수 $x$에 대하여 $x_n = \phi_p + \frac{x}{\sqrt{n}}$이라고 정의

`2` edf 변환

$$P\left(\sqrt{n}(\hat{\phi}_p - \phi_p) \le x\right) = P(\hat{\phi}_p \le x_n)$$

`3` 표본 분위수의 정의에 의해 $\hat{\phi}_p \le x_n$은 $\hat{F}_n(x_n) \ge p$와 동치이며, 이는 다시 표본 내에서 $x_n$보다 작거나 같은 관측치의 개수가 $np$ 이상임과 같음

$$P(\hat{\phi}_p \le x_n) = P\left(\sum_{i=1}^n I(X_i \le x_n) \ge np\right)$$

`4` $Y_i = I(X_i \le x_n)$라고 하면, $Y_i$는 성공 확률이 $p_n = F(x_n)$인 베르누이 분포를 따르고, 테일러 전개를 이용해, $p_n$을 근사하면

$$p_n = F\left(\phi_p + \frac{x}{\sqrt{n}}\right) \approx F(\phi_p) + \frac{x}{\sqrt{n}}f(\phi_p) = p + \frac{x}{\sqrt{n}}f(\phi_p)$$

`5` $\sum_{i=1}^n Y_i$의 기댓값은 $np_n$이고, 분산은 $np_n(1-p_n) \approx np(1-p)$입니다. 중심극한정리(CLT)에 의해 표준화된 합은 표준정규분포로 수렴

$$P\left(\sum_{i=1}^n Y_i \ge np\right) = P\left(\frac{\sum Y_i - np_n}{\sqrt{np_n(1-p_n)}} \ge \frac{np - np_n}{\sqrt{np_n(1-p_n)}}\right)$$

`6` 임계값에 극한을 취해 계산하면.

$$\frac{np - n\left(p + \frac{x}{\sqrt{n}}f(\phi_p)\right)}{\sqrt{np(1-p)}} = \frac{-x\sqrt{n}f(\phi_p)}{\sqrt{np(1-p)}} = \frac{-xf(\phi_p)}{\sqrt{p(1-p)}}$$

`7` 구하고자 하는 점근적 확률은

$$P\left(Z \ge \frac{-xf(\phi_p)}{\sqrt{p(1-p)}}\right) = P\left(Z \le \frac{xf(\phi_p)}{\sqrt{p(1-p)}}\right) = \Phi\left(x\frac{f(\phi_p)}{\sqrt{p(1-p)}}\right)$$

결론: $\sqrt{n}(\hat{\phi}_p - \phi_p) \xrightarrow{d} \mathcal{N}\left(0, \frac{p(1-p)}{f^2(\phi_p)}\right)$

---

### (b)

`1` $W_n$의 분포: $X_i \sim \mathcal{N}(\mu, \sigma^2)$이고 $W_n = \overline{X}_n$이므로, 표본평균의 정확한 분포는 $W_n \sim \mathcal{N}\left(\mu, \frac{\sigma^2}{n}\right)$

`2` $V_n$의 점근분포: (a)의 결과에서 $p=0.5$를 대입합니다. 정규분포의 중앙값은 $\phi_{0.5} = \mu$이고, 해당 지점에서의 확률밀도함수 값은 $f(\mu) = \frac{1}{\sqrt{2\pi\sigma^2}}$

- 점근 분산: $\frac{0.5(1-0.5)}{[f(\mu)]^2} = \frac{0.25}{1/(2\pi\sigma^2)} = \frac{\pi\sigma^2}{2}$

- 따라서 점근분포는 $\sqrt{n}(V_n - \mu) \xrightarrow{d} \mathcal{N}\left(0, \frac{\pi\sigma^2}{2}\right)$ 

ARE($V_n$ to $W_n$): 점근상대효율은 두 추정량의 점근분산의 비율로 계산됩니다 ($W_n$을 기준으로 할 때 $V_n$의 효율).

$$ARE(V_n, W_n) = \frac{\text{AsyVar}(W_n)}{\text{AsyVar}(V_n)} = \frac{\sigma^2}{\pi\sigma^2 / 2} = \frac{2}{\pi} \approx 0.6366$$

## EX 2

Suppose $X_{1},X_{2},...,X_{n}$ are Bernoulli(p), where $0<p<1$.

(a) Derive a $(1-\alpha)$ Wald confidence interval for p.

(b) Derive a $(1-\alpha)$ Wald confidence interval for $g(p)=\log\frac{p}{1-p}$ the log odds of p.

(c) Consider testing $H_{0}:p=p_{0}$ vs $H_{1}:p\ne p_{0}$. Derive the Wald test, the score test, and the LRT.  

## sol

### (a)

`1` 베르누이 분포에서 $p$의 최대우도추정량(MLE)은 $\hat{p} = \overline{X}_n$

`2` 중심극한정리에 의해 $\sqrt{n}(\hat{p} - p) \xrightarrow{d} \mathcal{N}(0, p(1-p))$를 따른다.

`3` Wald 방법론에서는 미지의 모수 $p$의 분산을 MLE로 대체한 표준오차 $SE(\hat{p}) = \sqrt{\frac{\hat{p}(1-\hat{p})}{n}}$를 사용

`결론`: $(1-\alpha)$ Wald 신뢰구간은 $\hat{p} \pm z_{\alpha/2} \sqrt{\frac{\hat{p}(1-\hat{p})}{n}}$

### (b)

`1` $g(p) = \log\left(\frac{p}{1-p}\right)$의 도함수를 구하면 $g'(p) = \frac{1}{p} + \frac{1}{1-p} = \frac{1}{p(1-p)}$

`2` 델타 방법(Delta Method)에 의해 $g(\hat{p})$의 점근분포를 도출

$$\sqrt{n}(g(\hat{p}) - g(p)) \xrightarrow{d} \mathcal{N}\left(0, [g'(p)]^2 \cdot p(1-p)\right) = \mathcal{N}\left(0, \frac{1}{p(1-p)}\right)$$

`3` 이에 대응하는 표준오차는 $SE(g(\hat{p})) = \frac{1}{\sqrt{n\hat{p}(1-\hat{p})}}$

`결론` : $(1-\alpha)$ Wald 신뢰구간은 $\log\left(\frac{\hat{p}}{1-\hat{p}}\right) \pm z_{\alpha/2} \frac{1}{\sqrt{n\hat{p}(1-\hat{p})}}$

### (c)

`1` Wald Test

- 추정치 $\hat{p}$의 분포를 기준으로 하며, 분산은 대립가설(MLE) 하의 값을 사용

-  검정통계량: $Z_W = \frac{\hat{p} - p_0}{\sqrt{\hat{p}(1-\hat{p})/n}}$

- 기각역: $|Z_W| > z_{\alpha/2}$ 일 때 $H_0$ 기각. (또는 $Z_W^2 > \chi^2_{1, \alpha}$)

`2` Score Test

- 귀무가설 $H_0$ 하에서의 정보량(Information)과 스코어(Score) 함수를 이용

- 로그 우도함수 $\log L(p) = \sum X_i \log p + (n - \sum X_i) \log(1-p)$의 스코어 함수와 피셔 정보량

$$U(p) = \frac{n(\hat{p} - p)}{p(1-p)}$, $\quad I(p) = \frac{n}{p(1-p)}$$

- 검정통계량: $Z_S = \frac{U(p_0)}{\sqrt{I(p_0)}} = \frac{\frac{n(\hat{p} - p_0)}{p_0(1-p_0)}}{\sqrt{\frac{n}{p_0(1-p_0)}}} = \frac{\hat{p} - p_0}{\sqrt{p_0(1-p_0)/n}}$

- 기각역: $|Z_S| > z_{\alpha/2}$ 일 때 $H_0$ 기각.

`3` Likelihood Ratio Test

- 귀무가설 $H_0$ 하에서의 최대 우도값과 전체 공간에서의 최대 우도값의 비를 사용

- 우도비 $\lambda = \frac{L(p_0)}{L(\hat{p})} = \left(\frac{p_0}{\hat{p}}\right)^{n\hat{p}} \left(\frac{1-p_0}{1-\hat{p}}\right)^{n(1-\hat{p})}$

- 검정통계량: $-2\log\lambda = 2n \left[ \hat{p} \log\left(\frac{\hat{p}}{p_0}\right) + (1-\hat{p}) \log\left(\frac{1-\hat{p}}{1-p_0}\right) \right]$

- 기각역: 점근적으로 카이제곱 분포를 따르므로, $-2\log\lambda > \chi^2_{1, \alpha}$ 일 때 $H_0$ 기각.
